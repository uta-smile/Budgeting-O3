# Boltz-2 1CLL TM-score benchmark (uv)

This notebook runs the local 1CLL benchmark from the repository's `uv` project. It is intentionally free of Colab magics, `sudo`, `pip install`, `/content` paths, and the Linux-only TMalign executable.

Start Jupyter from the project directory with:

```bash
uv sync
uv run --with jupyterlab --with ipykernel jupyter lab
```

The benchmark cell runs 100 stochastic Boltz-2 samples on CUDA and can take a while. The run is saved below `outputs/` and the later cells can be rerun without generating new structures by setting `RUN_BENCHMARK = False`.

````python
from pathlib import Path
import os
import shutil
import subprocess
import sys

import pandas as pd
from IPython.display import display


def find_project_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").is_file() and (candidate / "src" / "o3_boltz").is_dir():
            return candidate
    raise FileNotFoundError(
        "Could not find the o3-boltz2 project root. Start Jupyter from inside the repository."
    )


PROJECT_ROOT = find_project_root(Path.cwd()).resolve()
os.chdir(PROJECT_ROOT)

# Keep UV and Boltz's large downloads on the project filesystem.
os.environ.setdefault("UV_CACHE_DIR", str(PROJECT_ROOT / ".uv-cache"))
os.environ.setdefault("BOLTZ_CACHE", str(PROJECT_ROOT / "data" / "boltz_cache"))
UV = shutil.which("uv")
if UV is None:
    raise RuntimeError(
        "uv was not found on PATH. Launch this notebook with 'uv run ... jupyter lab'."
    )


def uv_run(*args: object, check: bool = True, capture_output: bool = False) -> subprocess.CompletedProcess[str]:
    """Run a command in this repository through the uv-managed environment."""
    command = [UV, "run", "--project", str(PROJECT_ROOT), *(str(arg) for arg in args)]
    return subprocess.run(
        command,
        cwd=PROJECT_ROOT,
        env=os.environ.copy(),
        text=True,
        check=check,
        capture_output=capture_output,
    )


print("Project:", PROJECT_ROOT)
print("Notebook kernel:", sys.executable)
print("uv:", UV)
````

````python
import torch

# This is a cheap package/kernelspec check; it does not load the model.
probe = uv_run("python", "-m", "boltz.main", "--help", capture_output=True)
print(probe.stdout.splitlines()[0])
print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("The generation cell requires a CUDA GPU; scoring-only cells can still read an existing run.")
````

````python
import yaml

REFERENCE_PDB = PROJECT_ROOT / "data" / "1CLL.pdb"
INPUT_YAML = PROJECT_ROOT / "data" / "1cll_boltz_input.yaml"
CONFIG_YAML = PROJECT_ROOT / "configs" / "1cll.yaml"

for required_file in (REFERENCE_PDB, INPUT_YAML, CONFIG_YAML):
    if not required_file.is_file():
        raise FileNotFoundError(f"Missing required project file: {required_file}")

input_data = yaml.safe_load(INPUT_YAML.read_text(encoding="utf-8"))
sequence = str(input_data["sequences"][0]["protein"]["sequence"]).strip().upper()
if len(sequence) != 144:
    raise ValueError(f"Expected the 144-residue PDB-observed 1CLL sequence, got {len(sequence)} residues")

print("Reference:", REFERENCE_PDB)
print("Input:", INPUT_YAML)
print("Sequence length:", len(sequence))
````

## Run the 100-sample benchmark

This calls the repository's built-in Best K-of-N runner through `uv`. It loads the vendored Boltz-2 adapter once, generates 100 samples with seed 0, writes PDBs, and scores them with the project's `tmtools` oracle normalized by the reference-chain length.

````python
N = 100
K = 10
SEED_START = 0
BUDGET_NAME = "n100_k10"
RUN_ID = f"notebook_n{N}_k{K}_seed{SEED_START}"
RUN_BENCHMARK = True

if (N, K) != (100, 10):
    raise ValueError("This notebook is wired to the project's n100_k10 budget.")

RUN_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "1cll"
    / "best_k_of_n"
    / BUDGET_NAME
    / "runs"
    / RUN_ID
    / f"seed_{SEED_START:04d}"
)
print("Run directory:", RUN_DIR)
````

````python
if RUN_BENCHMARK:
    if not torch.cuda.is_available():
        raise RuntimeError("Boltz-2 generation requires CUDA. Run this notebook on the lab GPU node.")

    uv_run(
        "python",
        "-m",
        "o3_boltz.cli",
        "--config",
        "configs/1cll.yaml",
        "--adapter",
        "adapters.boltz2_pfode:create",
        "--method",
        "best-k-of-n",
        "--only",
        BUDGET_NAME,
        "--replicates",
        "1",
        "--seed-start",
        str(SEED_START),
        "--run-id",
        RUN_ID,
    )
else:
    print("Skipping generation; using the existing run directory.")
````

````python
evaluation_path = RUN_DIR / "evaluations.csv"
if not evaluation_path.is_file():
    raise FileNotFoundError(
        f"No evaluation file at {evaluation_path}. Run the benchmark cell first or set RUN_ID to an existing run."
    )

raw_df = pd.read_csv(evaluation_path)
if "score" not in raw_df.columns:
    raise ValueError(f"Expected a score column in {evaluation_path}; found {list(raw_df.columns)}")

df = (
    raw_df.rename(columns={"score": "TM_score"})
    .sort_values("TM_score", ascending=False)
    .reset_index(drop=True)
)
df["rank"] = df.index + 1

print("Samples scored:", len(df))
display(df.head(K)[["rank", "TM_score", "structure"]])
````

````python
best_k_df = pd.DataFrame(
    {
        "K": range(1, len(df) + 1),
        "Best_TM_score": df["TM_score"].cummax().to_numpy(),
    }
)

print("Best-K summary:")
display(best_k_df[best_k_df["K"].isin([1, 5, 10, 20, 50, 100])])
print(f"Top-{K} mean TM-score: {df.head(K)['TM_score'].mean():.4f}")
print(f"Best TM-score: {df['TM_score'].max():.4f}")
````

````python
results_path = RUN_DIR / "1CLL_TMscore_results.csv"
df.to_csv(results_path, index=False)
print("Saved:", results_path)
display(df.head(10))
````

````python
# Plotting is optional because matplotlib is not part of the core uv lockfile.
try:
    import matplotlib.pyplot as plt
except ImportError:
    print("matplotlib is not installed; skipping the optional plot.")
else:
    ax = best_k_df.plot("K", "Best_TM_score", figsize=(7, 5), legend=False)
    ax.set_title("Boltz-2 Best-of-K on 1CLL")
    ax.set_ylabel("Best TM-score")
    ax.grid(True)
    plt.show()
````
